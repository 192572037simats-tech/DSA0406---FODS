## Experiment 1: Calculate subject-wise average marks and identify the highest-scoring subject

```python
import numpy as np

student_scores = np.array([
    [85, 78, 92, 74],
    [90, 88, 85, 80],
    [76, 82, 89, 91],
    [95, 91, 94, 86]
])

subjects = np.array(["Math", "Science", "English", "History"])
averages = np.mean(student_scores, axis=0)
highest_index = np.argmax(averages)

print("Average score for each subject:")
for subject, average in zip(subjects, averages):
    print(f"{subject}: {average:.2f}")
print(f"Highest average: {subjects[highest_index]} ({averages[highest_index]:.2f})")
```

## Experiment 2: Calculate the average product price from sales data

```python
import numpy as np

sales_data = np.array([
    [120.00, 125.00, 118.00],
    [75.00, 80.00, 78.00],
    [200.00, 190.00, 210.00]
])

average_price = np.mean(sales_data)
print(f"Average price of all products sold: ${average_price:.2f}")
```

## Experiment 3: Find the average sale price of houses with more than four bedrooms

```python
%%writefile houses.csv
bedrooms,square_footage,sale_price
3,1500,250000
5,2200,375000
6,2800,450000
4,1900,310000
5,2400,400000
```

```python
import numpy as np

house_data = np.genfromtxt("houses.csv", delimiter=",", skip_header=1)

more_than_four_bedrooms = house_data[:, 0] > 4
matching_prices = house_data[more_than_four_bedrooms, 2]

if matching_prices.size > 0:
    average_price = np.mean(matching_prices)
    print(f"Average sale price (>4 bedrooms): ${average_price:,.2f}")
else:
    print("No houses with more than four bedrooms were found.")
```

## Experiment 4: Calculate annual sales and percentage increase

```python
import numpy as np

sales_data = np.array([125000, 138000, 147500, 162000], dtype=float)

total_sales = np.sum(sales_data)
percentage_increase = ((sales_data[3] - sales_data[0]) / sales_data[0]) * 100

print(f"Total sales for the year: ${total_sales:,.2f}")
print(f"Increase from Q1 to Q4: {percentage_increase:.2f}%")
```

## Experiment 5: Calculate average fuel efficiency and percentage improvement

```python
import numpy as np

fuel_efficiency = np.array([28.5, 31.0, 35.5, 40.0, 33.0])
average_efficiency = np.mean(fuel_efficiency)

old_model = fuel_efficiency[0]
new_model = fuel_efficiency[1]
percentage_improvement = ((new_model - old_model) / old_model) * 100

print(f"Average fuel efficiency: {average_efficiency:.2f} mpg")
print(f"Improvement from model 1 to model 2: {percentage_improvement:.2f}%")
```

## Experiment 6: Calculate grocery purchase total with discount and tax

```python
item_prices = [2.50, 4.00, 3.25, 6.75]
quantities = [3, 2, 4, 1]
discount_rate = 10
tax_rate = 8

subtotal = sum(price * quantity
               for price, quantity in zip(item_prices, quantities))
discount_amount = subtotal * (discount_rate / 100)
discounted_total = subtotal - discount_amount
tax_amount = discounted_total * (tax_rate / 100)
final_total = discounted_total + tax_amount

print(f"Subtotal: ${subtotal:.2f}")
print(f"Discount: ${discount_amount:.2f}")
print(f"Tax: ${tax_amount:.2f}")
print(f"Total cost: ${final_total:.2f}")
```
